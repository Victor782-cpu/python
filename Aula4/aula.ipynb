```python
import pandas as pd

caminho = r"H:\Python\Python_3°etapa\Aula4\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"

df_cru = pd.read_excel(caminho, engine="openpyxl")

linha_cabecalho = df_cru[df_cru.iloc[:, 1].astype(str).str.contains("UF", na=False)].index[0]

df = pd.read_excel(caminho, skiprows=linha_cabecalho + 1, engine="openpyxl")

df.columns = df.columns.str.strip()

df = df.dropna(subset=["UF", "NOME DO MUNICÍPIO"])

col_2010 = [c for c in df.columns if "2010 (Alterações" in c][0]
col_2022 = [c for c in df.columns if "Censo 2022" in c][0]

df_estado = (
    df.groupby("UF")[[col_2010, col_2022]]
    .sum()
    .reset_index()
)

df_estado["CRESCIMENTO"] = df_estado[col_2022] - df_estado[col_2010]
df_estado_ordenado = df_estado.sort_values(by="CRESCIMENTO", ascending=False)

df_estado_ordenado.to_csv("populacao_por_estado_crescimento.csv", index=False)

df_municipio = df.copy()

df_municipio["CRESCIMENTO"] = df_municipio[col_2022] - df_municipio[col_2010]
df_municipio_ordenado = df_municipio.sort_values(by="CRESCIMENTO", ascending=False)

df_municipio_ordenado.to_csv("populacao_por_municipio_crescimento.csv", index=False)
```
